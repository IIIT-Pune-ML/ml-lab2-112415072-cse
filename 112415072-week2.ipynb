```python
X = np.array([ [10, 100, 5], [12, 110, 6], [11, 105, 7], [13, 500, 6], [14, 108, 8], [15, 115, 9] ]) 
print(np.mean(X,axis=0))
print(np.std(X,axis=0))
```

```python
condition=np.abs(X-np.mean(X,axis=0))>2*np.std(X,axis=0)
print(X[condition])
```
